```python
from ultralytics import YOLO

# Load a YOLO model directly (replace with the exact version/model name available in YOLO)
model = YOLO('yolo11s-seg.pt')  # YOLO's pre-defined lightweight model, for example

# Train the model
results = model.train(data=r'F:\ABDUL\ABDUL 2024\LANDSLIDE\data.yaml', 
                      epochs=6, 
                      imgsz=640, 
                      batch=16, 
                      workers=4, 
                      device='')  # Use 'cpu' if no GPU is available
```
